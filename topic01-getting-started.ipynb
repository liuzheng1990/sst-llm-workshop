ChatGPT is an example of LLM (Large Language Model). You can write commands to it, called "prompts", and let it respond to you.

There are a lot of things that ChatGPT can do, here are some examples:
1. Generate a [haiku](https://en.wikipedia.org/wiki/Haiku)
2. Improve the spelling and grammar of your text
3. Generate images (slow! Don't spam this!)
4. Ask it to come up with some puns
5. ... and many others!

Let's try the first one, generating a haiku. You can read more on the Wikipedia link above. In short, a haiku is a Japanese-style short poem that has three lines having the 5-7-5 syllable patterns.

Try giving this prompt to ChatGPT:

> Write a haiku about a group of students doing an LLM workshop in Singapore Polytechnic

See what haiku it gives you, and can note it down in the space below in between the triple backticks:

```
In Singapore's halls,
Students gather, minds take flight,
LLM dreams rise.
```

Then can press the check mark button on the top right of this cell, or you can press Ctrl+Enter to render the Markdown.

#### What's so interesting about ChatGPT generating a haiku?

You may think, "It's just three lines of text."
But take note on the number of syllables in each line.
Remember that a haiku in general must follow a 5-7-5 syllable structure.

Try counting the number of syllables in each line of your generated haiku above. See if it matches the 5-7-5 syllable structure.

As an example, here's the haiku ChatGPT generated for me:

```
Poly students learn
Language models softly hum
Future sparks bright paths
```

In this case, it does follow the 5-7-5 syllable structure, showing that ChatGPT can understand these that we might take for granted:

1. A haiku has a 5-7-5 syllable structure
2. It knows that "Poly students learn" consists of 5 syllables
3. It knows that "Language models softly hum" consists of 7 syllables
4. It knows that "Future sparks bright paths" consists of 5 syllables
5. It follows English grammatical structure
6. The theme matches our prompt of LLM workship in Singapore Polytechnic

Note that ChatGPT is not perfect. It might still make mistakes. Maybe for some of you it fails to do one of the 6 points above.

### 2. How does ChatGPT work?

Read more at the following page about [Tokens](https://aldrianobaja-sp.github.io/sst-sp-images/references/Topic%202%20-%20Deeper%20Dive%20into%20LLM/1.%20Tokens.html).

ChatGPT is a Large-language Model (LLM), and it works by predicting one word at a time (we call it "token" instead of word in LLM).

For example, after seeing the text "The cat likes to sleep in the", the LLM tries to predict what is the most likely word to come next.

![image.png](https://raw.githubusercontent.com/aldrianobaja-sp/sst-sp-images/refs/heads/main/2.%20Token%20Prediction.png)


### 3. 🛠️ Prompt ChatGPT to Solve Some Problems!
ChatGPT can help you with real problems like math, science, and planning tasks.  
Try these activities:  
- Ask ChatGPT: **“Solve this math problem: If a train travels 120 km in 2 hours, what is its speed?”**  
- Then ask: **“Explain the steps so I can understand how you got the answer.”**  
- Next, try a science question: **“Why does ice float on water? Explain in simple words.”**  
- Finally, ask a practical question: **“Create a short study schedule for 3 subjects in 2 hours.”**  

Notice how the answers change based on what you ask and how specific you are!


### 4. How to Improve Your Prompts to Get Better Results? (Prompt Engineering)
The way you ask a question matters!  
- Be **specific**: Instead of “Tell me about animals,” say “Explain why lions are called apex predators.”  
- Add **details**: If you want a short answer, say “Explain in 3 sentences.”  
Activity:  
- Write a vague prompt like “Tell me about space.”  
- Rewrite it to be more specific: “Explain why stars shine and give one example.”  
Compare the answers!

### 5. 🛠️ Improve Prompts and Compare!
Let’s practice:  
- First prompt: “Write about Charizard.”  
- Improved prompt: “Write a 5-sentence paragraph about Charizard, including its type, region, and best move.”  
Activity:  
- Try both prompts in ChatGPT and see which one gives you better information.  
- Discuss why the second prompt works better.
